**1.** Write a function `top_items(n, *sales)`. Each sale is a `(item, amount)` tuple, and you can pass in as many as you like. Return a list of the **names** of the `n` best-selling items, highest first. Use `sorted()` with a `lambda`.

Test your code with:
```python
print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print(top_items(5, ("pen", 50), ("book", 300)))
print(top_items(3))
```
Expected output:
```
['book', 'bag']
['book', 'pen']
[]
```
*Think:* what should happen when `n` is bigger than the number of sales, or when there are no sales at all?




````python

def top_items(n, *sales):
    sorted_sales = sorted(sales, key=lambda sale: sale[1], reverse=True)
    return [name for name, amount in sorted_sales[:n]]
print(top_items(2, ("pen", 50), ("book", 300), ("bag", 120)))
print(top_items(5, ("pen", 50), ("book", 300)))
````

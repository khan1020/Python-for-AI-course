```python
import matplotlib.pyplot as plt
```

```python
cars = ['BMW', 'Mercedes', 'Toyota', 'Honda', 'Ford']
years = [2010, 2012, 2015, 2018, 2020]
plt.plot(cars, years, marker='s', color='r')
plt.legend(["Car Models"])
plt.xlabel("Cars")
plt.ylabel("Years")
plt.title("Car Years")
plt.show()
plt.bar(cars, years)
plt.legend(["Car Models"])
plt.xlabel("Cars")
plt.ylabel("Years")
plt.title("Car Years")
plt.show()
```

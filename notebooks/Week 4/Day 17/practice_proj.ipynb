```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv('penguins.csv')
df.head()
```

```python
df.isna().sum()[df.isna().sum() > 0]
```

```python
df["bill_length_mm"] = df["bill_length_mm"].fillna(df["bill_length_mm"].median(), inplace=True)
```

```python
df["bill_length_mm"] = df["bill_length_mm"].fillna(df["bill_length_mm"].median(), inplace=True)
```

```python
df["flipper_length_mm"] = df["flipper_length_mm"].fillna(df["flipper_length_mm"].median(), inplace=True)
```

```python
df ["body_mass_g"] = df["body_mass_g"].fillna(df["body_mass_g"].median(),inplace = True)
```

```python
df.isna().sum()[df.isna().sum() > 0]
```

```python
df["bill_depth_mm"]= df["bill_depth_mm"].fillna(df["bill_depth_mm"].median(), inplace=True)
df.head()
```

```python
df["sex"] = df["sex"].map({"Male": 1, "Female": 0})
```

```python
df["sex"] = df["sex"].fillna(df["sex"].mode()[0], inplace=True)
```
